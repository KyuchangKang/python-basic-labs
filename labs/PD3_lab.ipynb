## Part G. 같은 실수의 다른 얼굴 — 이번엔 멈춥니다 (빈칸 없음)

이번에는 경영학과만 걸러낸 뒤 첫 사람을 이름표 `0` 으로 불러 봅니다.

```python
econ = df[df["학과"] == "경영"]
print("걸러낸 표의 이름표:", list(econ.index))

try:
    print(econ.loc[0])
except KeyError as e:
    print("KeyError:", e)

print("자리로:", econ.iloc[0]["이름"])
```

🔴 **같은 실수인데 이번에는 오류가 납니다.** 걸러낸 표에는 이름표 `0` 인 행이 **아예 없기** 때문입니다.
정렬한 표에서는 **남아 있어서 조용히 다른 사람**을 주었고, 여기서는 **없어서 멈춥니다.**
**멈추는 쪽이 오히려 다행입니다** — 틀린 답을 들고 가지 않으니까요.

## Part H. 슬라이스가 다릅니다 (빈칸 없음)

두 길은 **슬라이스에서도** 다릅니다. 줄 수를 세어 보세요.
마지막 줄은 **행과 열을 쉼표로 함께** 지정한 것입니다 — `loc` 은 열을 **이름**으로, `iloc` 은 열도 **자리**로 부릅니다.

```python
loc_slice = df.loc[0:2]
iloc_slice = df.iloc[0:2]
print("loc 0:2 ->", list(loc_slice["이름"]))
print("iloc 0:2 ->", list(iloc_slice["이름"]))
print("줄 수:", len(loc_slice), "vs", len(iloc_slice))

print("행과 열을 함께:", df.loc[0, "이름"], "/", df.iloc[0, 0])
```

🔴 **`loc` 의 슬라이스는 끝을 포함합니다.** 이름표를 **부르는** 것이라 끝 이름표까지 부릅니다.
`iloc` 은 P4의 리스트 슬라이스와 같아서 끝을 포함하지 않습니다.

## Part I. 불린 인덱싱 — 마스크를 눈으로 보기

열에 조건을 걸면 **값이 아니라 참/거짓이 줄줄이** 나옵니다. 이것을 **마스크**라고 부릅니다.
아래 빈칸에는 **90점 이상**을 고르는 비교 기호를 넣으세요.

```python
mask = df["점수"] ___ 90
print("마스크:", list(mask))

picked = df[mask]
print(picked)
print("몇 명:", mask.sum())
print("남은 이름표:", list(picked.index))
```

🔴 **남은 줄의 이름표를 보세요.** 걸러내도 이름표는 따라옵니다.

## Part J. 조건이 여럿일 때 — 괄호와 기호

파이썬의 `and`·`or`·`not` 이 아니라 **다른 기호 셋**을 씁니다.
아래는 **87점 이상이면서 경영학과**인 사람입니다 — 둘 다 참이어야 하는 기호를 넣으세요.

```python
both = df[(df["점수"] >= 87) ___ (df["학과"] == "경영")]
print(list(both["이름"]))

either = df[(df["학과"] == "경영") | (df["점수"] >= 91)]
print(list(either["이름"]))

nope = df[~(df["학과"] == "컴퓨터")]
print(list(nope["이름"]))
```

🔴 **조건마다 괄호**를 칩니다. 빼면 기호가 먼저 계산되어 오류가 납니다(Part L에서 봅니다).

## Part K. 목록으로 고르기와 값 바꾸기

값이 여럿일 때는 `|` 를 길게 잇는 대신 **목록 안에 있느냐를 묻는 메서드**를 씁니다.
이름 그대로 "안에 있나"를 뜻하는 다섯 글자입니다.
그리고 값을 바꿀 때는 **`.loc` 으로 한 번에** 씁니다 —
원본을 지키려고 **사본**에 씁니다(`df.copy()` · P4에서 리스트를 복사하던 그것과 같은 뜻입니다).

```python
chosen = df[df["이름"].___(["강수빈", "한도윤"])]
print(list(chosen["이름"]))

copy_df = df.copy()
copy_df.loc[copy_df["점수"] >= 90, "점수"] = 100
print(list(copy_df["점수"]))
print("원본은 그대로:", list(df["점수"]))
```

🔴 **대괄호를 두 번 이어 쓰지 마세요** — `copy_df[조건]["점수"] = 100` 처럼 쓰면
**원본이 안 바뀌거나 경고가 뜹니다.** `df.loc[조건, "열"] = 값` 한 줄이 규칙입니다.

## Part L. 오류 읽기 — 고르고 거를 때 (빈칸 없음)

다섯 가지를 일부러 내 봅니다. **오류의 이름**이 무엇을 헷갈렸는지 알려 줍니다.

```python
try:
    econ.loc[0]
except KeyError as e:
    print("없는 이름표 ->", type(e).__name__, e)

try:
    df.loc[0, "등수"]
except KeyError as e:
    print("없는 열 ->", type(e).__name__, e)

try:
    df.iloc[9]
except IndexError as e:
    print("표 밖의 자리 ->", type(e).__name__)
```

```python
try:
    df[df["점수"] >= 87 & df["학과"] == "경영"]
except TypeError as e:
    print("괄호를 빼먹었다 ->", type(e).__name__)

try:
    df[(df["점수"] >= 87) and (df["학과"] == "경영")]
except ValueError as e:
    print("and 를 썼다 ->", type(e).__name__)
```

🔴 **막히면 `print(list(df.index))`** — 오늘 사고는 전부 거기서 드러납니다.
**괄호 오류의 메시지 문구는 판다스 버전마다 다릅니다.** 이름만 기억하세요.
