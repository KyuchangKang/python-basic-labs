## Part F. 세트피스 — 설문 응답 정제 ⭐

**⚠️ 실행하기 전에 손으로 먼저 세어 보세요.** 파이썬이라고 답한 사람은 몇 명인가요?

아래 코드는 배운 대로 공백을 없애고(`strip`) 소문자로 바꾼(`lower`) 뒤 비교합니다.
**일부러 이대로 두었습니다.** 고치지 말고 그냥 실행해 보세요.

```python
answers = ["  Python ", "python", "PYTHON  ", " java", "Python"]
count = 0
for a in answers:
    a.strip()              # 양끝 공백을 없앤다
    a.lower()              # 소문자로 바꾼다
    if a == "python":
        count = count + 1
wrong_count = count
print(wrong_count)
```
