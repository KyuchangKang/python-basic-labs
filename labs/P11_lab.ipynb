## Part G. 두 방식을 그대로 다시 돌립니다

> `.to_dict()`는 판다스 결과를 **우리가 아는 딕셔너리로** 되돌리는 것입니다(나란히 비교하기 편하려고요).


```python
with open("scores_blank.csv", encoding="utf-8") as f:
    lines2 = f.readlines()

total2 = {}
count2 = {}
for line in lines2[1:]:
    name, dept, score = line.strip().split(",")
    if score == "":
        point = 0
    else:
        point = int(score)
    total2[dept] = total2.get(dept, 0) + point
    count2[dept] = count2.get(dept, 0) + 1

py_avg = {}
for d in total2:
    py_avg[d] = round(total2[d] / count2[d], 1)

py_rank = sorted(py_avg, key=py_avg.get, reverse=True)
print("순수 파이썬:", py_avg, "· 1위:", py_rank[0])

df2 = pd.read_csv("scores_blank.csv")
pd_avg = df2.groupby("학과")["점수"].mean().round(1).to_dict()
pd_rank = sorted(pd_avg, key=pd_avg.get, reverse=True)
print("판다스     :", pd_avg, "· 1위:", pd_rank[0])
```

## Part H. 왜 갈렸나 — `len()`과 `count()`

우리 코드는 `if score == "": point = 0`이라고 **내가 정했습니다.**
판다스는 빈 칸을 **`NaN`**(값이 없음)으로 읽고 `mean()`에서 **빼고** 계산합니다.

> 아래 마지막 줄의 `df2[df2["학과"] == "컴퓨터"]`는 **컴퓨터학과 줄만 골라낸 것**입니다.
> 이렇게 조건으로 골라내는 법은 **2학기 PD3**에서 배웁니다 — 지금은 그렇게만 알아 두세요.


```python
print("판다스가 읽은 점수:", list(df2["점수"]))
print("줄 수 len(df2):", len(df2))
print("값이 있는 칸 count():", df2["점수"].count())
print("컴퓨터 합계는 양쪽 다:", total2["컴퓨터"])
print("나눈 수 — 파이썬:", count2["컴퓨터"], "/ 판다스:", df2[df2["학과"] == "컴퓨터"]["점수"].count())
```

**합은 같고 나누는 수가 다릅니다.**

그럼 어느 쪽이 맞을까요? **둘 다 맞을 수 있습니다.**
시험을 안 본 것이 0점 처리라면 60.0이, 연기된 것이라면 90.0이 맞습니다.

> **파이썬이 정할 문제도, 판다스가 정할 문제도 아닙니다.** 데이터를 아는 사람이 정할 문제입니다.
> 다만 순수 파이썬에서는 그 결정이 **한 줄로 눈에 보이고**, 판다스 한 줄에서는 **보이지 않습니다.**
> **도구는 코드를 줄여 주지만 판단을 대신해 주지 않습니다.**
