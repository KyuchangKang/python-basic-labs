## Part G. 범인, 그리고 고치기 — `key=`로 기준을 말해 주기

**점수 여섯 개가 아니라 이름 여섯 개였습니다.**
P6에서 배웠지요 — **딕셔너리를 순회하면 나오는 것은 열쇠**입니다. `sorted`도 순회해서 꺼냅니다.
그래서 파이썬은 **이름을 가나다 역순으로** 줄 세웠습니다. 'ㅎ'인 한도윤이 맨 앞, 'ㄱ'인 김하늘이 맨 뒤로요.

> `sorted`는 틀리지 않았습니다. 우리가 **무엇을 기준으로 줄 세울지 말해 주지 않았을 뿐입니다.**
> **말해 주지 않은 기준은 파이썬이 정합니다.**

딕셔너리에는 **이름을 주면 점수를 돌려주는 메서드**가 이미 있습니다 — P6에서 "없으면 기본값"과 함께 썼던 그것입니다.

## Part H. 람다 · `max` · `min`

**손으로 구한 것과 같아졌습니다.** ✅

기준이 이미 있는 함수가 아닐 때는 **그 자리에서 한 줄 함수**를 만듭니다. 그것이 **람다**입니다.
튜플은 **첫 원소부터** 비교하므로, 그냥 정렬하면 세트피스와 똑같이 **이름으로** 줄을 섭니다.


```python
rows = [("김하늘", 92), ("이서준", 88), ("정민아", 85), ("박지호", 85)]

print(sorted(rows))
print(sorted(rows, key=lambda r: r[1], reverse=True))
```

**동점을 보세요.** 정민아와 박지호가 둘 다 85점인데 **정민아가 앞**입니다.
파이썬 정렬은 **동점이면 원래 순서를 그대로 지킵니다** — **안정 정렬**이라고 합니다.

같은 `key=`가 `max`·`min`에도 그대로 들어갑니다. 🔴 `max(scores)`도 **오류가 나지 않습니다.**


```python
print("max(scores):", max(scores))
print("max(scores, key=scores.get):", max(scores, key=scores.get))
print("min(scores, key=scores.get):", min(scores, key=scores.get))
```
