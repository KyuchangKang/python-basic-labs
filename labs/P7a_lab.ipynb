## Part C. `return`은 즉시 함수를 끝낸다

두 번째 함수는 60 이상인 값을 찾습니다. `[48, 55, 72, 90]`에는 60 이상이 **둘**인데,
결과는 몇이 나올까요? **실행 전에 먼저 예측해 보세요.**


````python
def grade(score):
    if score >= 60:
        return "합격"
    return "불합격"        # 위에서 끝나면 여기는 오지 않는다

print(grade(60), grade(59))

def first_big(nums):
    for n in nums:
        if n >= 60:
            return n       # 찾자마자 함수가 끝난다
    return -1              # 하나도 못 찾았을 때

found = first_big([48, 55, 72, 90])
print(found)               # 90 일까 72 일까
````

## Part E. 세트피스 — 합격자 세기 ⭐

여섯 학생의 점수입니다. **60점 이상이 합격**입니다(P2의 그 경계 — 60은 합격).

```
90, 55, 72, 60, 48, 83
```

**⚠️ 실행하기 전에 손으로 먼저 세어 보세요.** 합격은 몇 명인가요?

아래는 판정을 함수로 묶어 세는 코드입니다. **고치지 말고 그냥 실행해 보세요.**


````python
scores = [90, 55, 72, 60, 48, 83]

def is_pass(score):
    if score >= 60:
        print("합격")
    else:
        print("불합격")

wrong_passed = 0
for s in scores:
    if is_pass(s):                          # 합격이면 한 명 더
        wrong_passed = wrong_passed + 1
print(wrong_passed)
````

> **손으로 센 답은 `4`명이었습니다.** 그리고 화면에 찍힌 판정 여섯 줄도 **전부 맞습니다.**
>
> 그런데 마지막 숫자는 어떻게 나왔나요? **오류는 한 줄도 나지 않았습니다.**
> 화면은 맞는데 숫자만 틀린 이유를 Part F에서 확인합니다.


## Part F. 왜 그런가 — 함수를 직접 찍어 보기

`is_pass(90)`을 `print`로 감싸면 **두 줄**이 나옵니다.
어느 줄이 함수 안의 `print`가 쓴 것이고, 어느 줄이 **돌려준 값**인가요?


````python
print(is_pass(90))              # 두 줄이 나온다

returned = is_pass(72)          # 화면에 한 줄 더 찍힌다
print(returned is None)         # 돌려준 값은 None 인가

print(bool(None), bool(True))   # if 는 None 을 참으로 볼까 거짓으로 볼까
````

## Part G. 고치는 법 — 돌려준다

바꿀 것은 **한 줄**입니다. `print` 대신 `return`으로 비교 결과를 돌려주세요.

📌 빈칸 4 — **60점도 합격**입니다. P2에서 배운 그 경계를 기억하세요.


````python
def is_pass_ok(score):
    return score ___ 60          # 빈칸 4 — 60 도 합격이 되도록

print(is_pass_ok(90), is_pass_ok(55), is_pass_ok(60))

right_passed = 0
for s in scores:
    if is_pass_ok(s):
        right_passed = right_passed + 1
print(right_passed)
````

**보여 주는 일은 부르는 쪽이 합니다.** 함수는 판정만 하고, 화면 출력은 밖에서 하세요.


````python
for s in scores:
    if is_pass_ok(s):
        print(s, "합격")
    else:
        print(s, "불합격")
````

````python
try:
    add(2)                       # 인자를 덜 넣으면
except TypeError as e:
    print("TypeError:", e)

try:
    print(len(shout("hi")))      # None 에 len 을 쓰면
except TypeError as e:
    print("TypeError:", e)

def count_pass_bug(nums):
    n = 0
    for x in nums:
        if x >= 60:
            n = n + 1
        return n                 # ← return 이 for 안에 있다
bug = count_pass_bug(scores)
print(bug)                       # 4 가 나올까
````
