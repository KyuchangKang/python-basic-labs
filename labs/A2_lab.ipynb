## Part A. A1의 `Student` — 가산점 자리와 `__str__`

A1에서 만든 설계도에 **가산점(`bonus`)** 자리를 두고, `print` 모습을 꾸미는 `__str__`을 더했습니다.
`__init__`·`add`·`average`는 A1 그대로이고, 평균에 `self.bonus`가 더해지는 것만 달라졌습니다.
이 셀은 **그대로 실행만 하세요.**

```python
class Student:
    def __init__(self, name, dept):
        self.name = name
        self.dept = dept
        self.scores = []
        self.bonus = 0

    def add(self, point):
        self.scores.append(point)

    def average(self):
        return round(sum(self.scores) / len(self.scores) + self.bonus, 1)

    def __str__(self):
        return self.name + "(" + self.dept + ")"

subin = Student("강수빈", "컴퓨터")
minjun = Student("노민준", "경영")

for p in [92, 87, 88]:
    subin.add(p)
for p in [78, 82, 80]:
    minjun.add(p)

print(subin)
print(minjun)
```

**`print`가 화면에 찍기 직전에 파이썬이 `__str__`을 부릅니다.** 내가 부르지 않습니다 — `__init__`과 같은 방식입니다.
A1에서 `<__main__.Student object at 0x...>`가 나온다고 하고 넘어간 그 자리를 고친 것입니다.

> **빈칸 3** — 리스트·딕셔너리 **안에서** 파이썬이 부르는 특수 메서드의 이름을 쓰세요.

```python
class StudentR(Student):
    def ___(self):
        return self.__str__()

r1 = StudentR("강수빈", "컴퓨터")
r2 = StudentR("노민준", "경영")

print([str(s) for s in [subin, minjun]])
print([r1, r2])
```

## Part F. 세트피스 ⭐ — 장학생을 만들어 돌려 봅니다

장학생은 **가산점을 받아서** 만들어지므로 `__init__`을 덮어씁니다.
**읽어 보면 아주 자연스러운 코드입니다.**

```python
class ScholarStudent(Student):
    def __init__(self, name, dept, bonus):
        self.bonus = bonus
        super().__init__(name, dept)

    def __str__(self):
        return "★ " + super().__str__()

haram = Student("정하람", "컴퓨터")
for p in [93, 91, 92]:
    haram.add(p)

harin = ScholarStudent("오하린", "컴퓨터", 5)
for p in [84, 88, 86]:
    harin.add(p)

doyun = ScholarStudent("한도윤", "경영", 5)
for p in [85, 89, 87]:
    doyun.add(p)

everyone = [subin, minjun, haram, harin, doyun]
for s in everyone:
    print(s, s.average())

print("장학금 명단:", [s.name for s in everyone if s.average() >= 90.0])
```

**손으로 구한 값과 같은가요?**

🔴 그런데 **별표는 제대로 붙었습니다.** `__str__` 오버라이드도 `super()`도 잘 돌았어요.
**가산점이 없는 세 학생의 값도 전부 맞습니다.** 명단도 비어 있지 않고 한 명이 들어 있고요.

## Part H. 실행 순서 — `ScholarStudent(...)` 한 줄이 하는 일

`ScholarStudent("오하린", "컴퓨터", 5)` 한 줄은 실제로 이렇게 흘러갑니다.

| 순서 | 실행되는 곳 | 하는 일 | 이때 `bonus` |
|---|---|---|---|
| 1 | 자식의 `__init__` 시작 | — | (없음) |
| 2 | `self.bonus = bonus` | 가산점을 넣는다 | **5** |
| 3 | 자식이 부모를 부르는 줄 | **부모의 `__init__`을 지금 실행** | 5 |
| 4 | `Student.__init__` 안 | `name`·`dept`·`scores` | 5 |
| 5 | `Student.__init__` 안 | **`self.bonus = 0`** ← 여기 | **0** |
| 6 | 부모 끝 → 자식 끝 | — | **0** |

**부모는 자식이 먼저 적어 둔 값을 모릅니다.** 그냥 자기 할 일을 할 뿐이고, 그중에 `self.bonus = 0`이 있습니다.
**나중에 실행된 대입이 이깁니다** — P1에서 세 자리를 돌려 앉힐 때 배운 그 성질입니다.

## Part J. 오류 읽기 — 상속을 쓸 때

```python
first_error = ___

class NoSuper(Student):
    def __init__(self, name, dept, bonus):
        self.bonus = bonus

try:
    NoSuper("가", "나", 5).average()
    caught1 = "없음"
except Exception as e:
    caught1 = type(e).__name__
print("1)", caught1, "/ 내 짐작:", first_error)

try:
    ScholarStudentFixed("오하린", "컴퓨터")
except TypeError as e:
    print("2)", type(e).__name__, "->", e)

class BadSuper(Student):
    def __init__(self, name, dept, bonus):
        super().__init__(name)
        self.bonus = bonus

try:
    BadSuper("가", "나", 5)
except TypeError as e:
    print("3)", type(e).__name__, "->", e)

class SelfToSuper(Student):
    def __init__(self, name, dept, bonus):
        super().__init__(self, name, dept)
        self.bonus = bonus

try:
    SelfToSuper("가", "나", 5)
except TypeError as e:
    print("4)", type(e).__name__, "->", e)
```
