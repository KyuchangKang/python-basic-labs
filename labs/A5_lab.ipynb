## Part I. 왕복하면 달라지는 것

JSON 은 파이썬이 아니라 **그냥 텍스트** 입니다. 그래서 왕복하면 조용히 달라지는 것이 있습니다.
아래에서 **열쇠가 무엇이 되어 돌아오는지** 보고, 그 모양을 **그대로** 적으세요.
튜플·`True`·`None`·집합이 어떻게 되는지도 함께 봅니다.
가운데 `json.dumps` 에는 **한글 옵션을 일부러 빼 두었습니다** — 빼면 화면이 어떻게 되는지도 함께 보세요.

````python
year_count = {1: "일학년", 2: "이학년"}
back = json.loads(json.dumps(year_count))
print(back)
print(back.get(1, "못 찾음"))

expected_key = ___      # 위 back 의 첫 번째 열쇠를 그대로 적으세요
print(expected_key in back)

box = {"동아리": ("코딩", "사진"), "확정": True, "비고": None}
text = json.dumps(box)
print(text)
print(type(json.loads(text)["동아리"]).__name__)

try:
    json.dumps({"명단": {"강수빈", "노민준"}})
except TypeError as e:
    print("TypeError:", e)
````

🔴 **튜플은 리스트가 되고, `True` 는 `true`, `None` 은 `null` 이 되고, 집합은 아예 담기지 않습니다.**
그리고 **열쇠는 언제나 문자열** 이 됩니다 — 학번·연도를 열쇠로 쓴 데이터는 왕복 뒤에 숫자로 찾으면 못 찾습니다.

## Part J. 겹겹을 표로 펴기 — PD1로 가는 다리

겹겹인 데이터는 보기엔 좋지만 집계하기엔 불편합니다.
**한 줄이 한 사람인 표** 로 펴 두면 한 겹만 세도 맞습니다.
표의 한 줄은 딕셔너리 하나, 표 전체는 그 딕셔너리들의 **리스트** 입니다.

````python
rows = []
for booth, grades in festival["부스"].items():
    for grade, members in grades.items():
        for name in members:
            rows.append({"이름": name, "부스": booth, "학년": grade})

for row in rows[:3]:
    print(row)
print("줄 수:", len(rows))
````

**이제 `len(rows)` 가 사람 수입니다** — 한 겹으로 폈으니까요.
세트피스가 틀린 이유를 뒤집어 말하면 이렇습니다: **셀 것을 먼저 펴 놓으면 실수할 자리가 없어집니다.**

> **판다스가 받는 것이 바로 이 모양** 입니다 — 딕셔너리들의 리스트.
> 다음 시간 **PD1** 에서 같은 방식으로 편 표를 판다스에 넣습니다.
