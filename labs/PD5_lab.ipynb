## Part I. 값마다 내 함수를 통과시키기

곱하기·더하기로 안 되는 규칙은 **함수로 적어 통과시킵니다.**
**빈칸 둘 다에** **값을 하나씩 꺼내 함수에 넣고 결과를 모아 주는 메서드**를 넣으세요(같은 메서드입니다).
아래쪽은 **람다**(P9에서 배운 이름 없는 한 줄 함수)로 같은 규칙을 쓴 것입니다.
🔴 넘기는 것은 **함수 자체**입니다 — `grade` 이지 `grade()` 가 아닙니다(P9 §6.2).

```python
def grade(score):
    if score >= 90:
        return "A"
    return "B"

df["등급"] = df["점수"].___(grade)
grades = df["등급"].tolist()
print("함수로:", grades)

grades_lambda = df["점수"].___(lambda s: "A" if s >= 90 else "B").tolist()
print("람다로:", grades_lambda)
```

**두 줄의 결과가 같습니다** — 같은 규칙을 이름 붙여 쓰느냐 한 줄로 쓰느냐의 차이일 뿐입니다.
🔴 **곱셈 하나로 되는 일에는 쓰지 마세요.** `df["점수"] * 1.1` 이 더 짧고 빠릅니다.

## Part J. 대응표로 바꾸기와 문자열 도구

**앞의 두 빈칸에는 같은 메서드**를 넣으세요 — **딕셔너리를 주면 값을 갈아 끼워 주는** 것입니다(P6의 그 딕셔너리).
한 번은 **고친 표**에, 한 번은 **고치기 전 표**에 겁니다. 결과가 어떻게 다른지 보세요.
마지막 두 줄은 **그 글자가 들어 있는지** 묻는 `.str.contains()` 입니다 — 채우지 않아도 됩니다.

```python
df["학과코드"] = df["학과"].___({"컴퓨터": "CS", "경영": "BA"})
codes = df["학과코드"].tolist()
print("고친 표에 걸면:", codes)

codes_dirty = dirty["학과"].___({"컴퓨터": "CS", "경영": "BA"})
print("고치기 전 표에 걸면:", codes_dirty.tolist())
print("대응표에 없어 비게 된 칸:", int(codes_dirty.isnull().sum()), "개")

has_cs = dirty["학과"].str.contains("컴퓨터")
print("컴퓨터가 들어 있나:", has_cs.tolist(), "->", int(has_cs.sum()), "명")
```

🔴 **대응표에 없는 값은 조용히 비게 됩니다** — 오류가 아닙니다(PD4의 그 `NaN`).
고치기 전 표에서는 **한 사람만** 그렇게 되지요. 오늘의 사고가 **또 다른 얼굴로** 나타난 것입니다.

🔴 **마지막 줄을 보세요.** 그 메서드로 걸렀다면 **공백이 있어도 걸러졌을** 것입니다.
다만 이것은 **증상을 피한 것**이지 고친 것이 아닙니다 — 값은 여전히 `"컴퓨터 "` 입니다.

## Part K. 오류 읽기 — 네 가지 (빈칸 없음)

**오류의 이름**이 무엇을 헷갈렸는지 알려 줍니다.

```python
try:
    df["학과 "]
except KeyError as e:
    print("열 이름 뒤에 공백 ->", type(e).__name__, e)

try:
    df["점수"].str.lower()
except AttributeError as e:
    print("숫자 열에 문자열 도구 ->", type(e).__name__)

try:
    df["점수"].apply(grade())
except TypeError as e:
    print("괄호를 붙여 grade() 를 먼저 불러 버렸다 ->", type(e).__name__)
```
