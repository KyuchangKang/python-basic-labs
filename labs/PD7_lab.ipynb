## Part A. 오늘의 표 셋

점수표 한 장과 학과 명단 두 장입니다. **점수표에는 학과 열이 없습니다** — 그래서 붙이는 일이 필요합니다.

```python
scores = pd.DataFrame({
    "이름": ["강수빈", "노민준", "정하람", "오하린", "한도윤"],
    "점수": [89, 80, 92, 91, 87],
})
print(scores)
print("줄 수:", len(scores), "/ 반 평균:", scores["점수"].mean())

cs = pd.DataFrame({"이름": ["정하람", "오하린"], "학과": ["컴퓨터", "컴퓨터"]})
ba = pd.DataFrame({"이름": ["노민준", "한도윤"], "학과": ["경영", "경영"]})
print(cs)
print(ba)
```

## Part B. 명단 두 장을 한 장으로

열 이름이 양쪽 같으니 **아래로 이어 붙일** 수 있습니다.
🔴 표는 **대괄호에 담아 하나로** 넘깁니다 — 나란히 두 개 쓰면 오류입니다.

```python
depts_raw = pd.___([cs, ba])
print(depts_raw)
print("줄 수:", len(cs), "+", len(ba), "=", len(depts_raw))
```

## Part C. 이름표가 겹쳤습니다

위 출력의 **왼쪽 번호**를 보세요. 각 표가 **자기 이름표를 그대로 가지고 왔습니다.**
그래서 아래 첫 줄은 한 줄이 아니라 **두 줄**을 냅니다(PD3에서 배운 그 이름표).

붙일 때 **번호를 새로 매기라는 옵션**을 주면 됩니다 — 이름 그대로 "인덱스를 무시하라"는 뜻입니다.

## Part E. 검산 — 세 숫자를 나란히

붙이기 **전** 두 표의 줄 수와 붙인 **뒤**의 줄 수를 나란히 찍습니다.

줄 수가 어긋났다면 **누가 빠졌는지** 찾습니다(둘째 셀). PD3의 `~`(아닌 것)를 그대로 쓰고,

```python
print(len(scores), len(depts), len(result))
print("붙인 표의 평균:", result["점수"].mean())
print("진짜 반 평균  :", scores["점수"].mean())
```

## Part F. 왼쪽 표를 한 줄도 버리지 않으려면

`how=` 는 **누구를 남길 것인가**를 정하는 자리입니다. 왼쪽 표를 다 남기라고 말해 주세요.

```python
kept = pd.merge(scores, depts, on="이름", how=___)
print(kept)
print("줄 수:", len(kept), "/ 평균:", kept["점수"].mean())
```

고쳤으니 이제 학과별 평균을 내 봅시다. PD6에서 배운 그대로입니다.
🔴 **인원 합을 줄 수와 맞춰 보세요.** 무엇이 보이나요?

```python
g = kept.groupby("학과")["점수"].agg(["count", "mean"])
print(g)
print("인원 합:", g["count"].sum(), "/ 줄 수:", len(kept))
```

## Part G. 반대 얼굴 — 줄이 늘어난다

명단이 틀린 것이 아닙니다 — 정말로 두 학과에 속합니다.

🔴 아래 둘째 줄은 **서로 다른 값이 몇 가지인지** 세는 것입니다(PD5에서 쓴 그것).

```python
depts2 = pd.concat([depts, pd.DataFrame({"이름": ["오하린"], "학과": ["경영"]})],
                   ignore_index=True)
print(depts2)
kinds = depts2["이름"].___()
print("명단 줄 수:", len(depts2), "/ 이름 가짓수:", kinds)
```

```python
left2 = pd.merge(scores, depts2, on="이름", how="left")
print(left2)
print("줄 수:", len(left2), "/ 평균:", round(left2["점수"].mean(), 2))
```

## Part H. 검산이 통과해 버릴 때
