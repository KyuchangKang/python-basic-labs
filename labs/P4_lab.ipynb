## Part B. 뒤에서 세기 · 없는 칸을 꺼내면

`-1`이 마지막입니다. **길이를 몰라도** 마지막을 꺼낼 수 있습니다.
없는 칸을 꺼내면 어떤 오류가 나는지도 직접 봅니다(일부러 내는 오류라 `try`로 감쌌습니다).

## Part C. 슬라이스 — 여러 칸을 한 번에

읽는 법은 **"시작은 포함, 끝은 직전까지"** 입니다. P3a의 `range(5)`가 `0, 1, 2, 3, 4`였던 것과 같은 규칙입니다.

📌 `[3, 5, 4]`를 얻으려면 **어디서 멈춰야** 할까요? 값이 아니라 **경계** 번호를 생각하세요.

```python
print(plan[1:3])           # 1부터 3 직전까지
print(plan[:2])            # 처음부터
print(plan[3:])            # 끝까지
print(plan[:-1])           # 마지막 직전까지
mid = plan[1:___]          # 빈칸 2 — [3, 5, 4] 가 나오려면?
print(mid)
```

```python
print(plan[2:2])                  # 시작과 끝이 같으면
print(plan[3:1])                  # 거꾸로 잘라도 오류가 아니다
print(plan[3:99])                 # 범위를 넘겨도 있는 만큼만
halves = plan[:2] + plan[2:]      # 두 조각을 이으면 원래대로
halves_ok = (halves == plan)
print(halves, halves_ok)
```

## Part D. 바꾸기 — 리스트는 고칠 수 있다

`remove`는 **값**으로, `pop`은 **자리**(인덱스)로 지웁니다. `pop`은 꺼낸 값을 돌려주기도 합니다.

```python
p = [2, 3, 5, 4, 6]
p.append(7)                # 맨 뒤에 덧붙이기
print(p)
p.insert(0, 1)             # 0번 자리에 끼워넣기
print(p)
p.remove(5)                # 값 5를 지우기
print(p)
taken = p.pop()            # 맨 뒤를 꺼내며 지우기
print(taken, p)
```

```python
plan = [2, 3, 5, 4, 6]                          # 다시 원래대로
print(len(plan), sum(plan), max(plan), min(plan))
print(5 in plan, 7 in plan)                     # 들어 있나
plan_sum = sum(plan)
print(plan_sum)
```

```python
plan = [2, 3, 5, 4, 6]     # 합계 20
backup = ___               # 빈칸 3 — 고치기 전에 백업해 두자
plan[4] = 9                # 5주차를 9문제로
print(plan)
print(backup)
plan_total = sum(plan)
backup_total = sum(backup)
print(plan_total, backup_total)
```

## Part F. `id()`로 검산 — 리스트가 둘일까, 하나일까

`id()`는 값이 아니라 **그 값이 놓인 자리의 번호표**를 돌려줍니다.
두 이름이 **같은 리스트**를 보고 있는지 물어봅시다.

📌 빈칸 4에는 "같은가?"를 묻는 **비교 연산자**가 들어갑니다(P2에서 배웠습니다).

```python
same = (id(plan) ___ id(backup))   # 빈칸 4 — 같은 리스트인가?
print(same)
plan = backup                      # 백업으로 되돌려 보자
restored_total = sum(plan)
print(restored_total)              # 되돌아갔을까?
```

## Part G. 고치는 법 — 잘라서 새로 만든다

슬라이스는 꺼낸 값들로 **새 리스트를 만들어** 돌려줍니다. 그래서 원본과 **다른 리스트**가 됩니다.

📌 빈칸 5 — **설명서 §4.2 슬라이스 표**에서 **"전부"를 가져오던 그 표기**를 떠올리세요.

```python
plan = [2, 3, 5, 4, 6]        # 원래대로 다시 준비
safe = plan[___]              # 빈칸 5 — 통째로 잘라 새 리스트를 만든다
plan[4] = 9                   # 아까와 똑같이 고친다
safe_total = sum(safe)
print(sum(plan), safe_total)
print(id(plan) == id(safe))   # 다른 리스트인가?
```

```python
plan = [2, 3, 5, 4, 6]
c1 = plan[:]                  # 슬라이스
c2 = list(plan)               # list() 로 새로 만들기
c3 = plan.copy()              # 이름이 가장 분명하다
plan[4] = 9
print(sum(c1), sum(c2), sum(c3))   # 셋 다 같은 일을 한다
```

## Part H. ⑤확인 — 리스트에서 나는 오류 3종

일부러 오류를 내고 **메시지를 읽는** 연습입니다. `try`로 감쌌으니 노트북은 멈추지 않습니다.
(`try`/`except` 문법은 **P8에서 배웁니다.** 지금은 메시지만 읽으세요.)

```python
p = [2, 3, 5, 4, 6]
try:
    p[5]                   # 마지막은 4인데 5를 달라고 했다
except IndexError as e:
    print("IndexError:", e)
try:
    p["1"]                 # 인덱스에 따옴표를 붙였다
except TypeError as e:
    print("TypeError:", e)
try:
    p.add(7)               # append 를 add 로 잘못 기억했다
except AttributeError as e:
    print("AttributeError:", e)
```
