## Part I. 고치기

> **빈칸 7** — 판다스에서 **값을 숫자로 바꾸는 함수**의 이름을 쓰세요(`pd.` 다음에 오는 것).
> `numeric`이라는 낱말이 들어갑니다.

```python
sup["보충"] = pd.___(sup["보충"], errors="coerce")

fixed = sup.sort_values("보충", ascending=False)

print(fixed)
print("고친 뒤 1등:", fixed["이름"].iloc[0], fixed["보충"].max())
print("손으로 구한 1등:", expected_name, expected_score)
print()
print("평균 :", sup["보충"].mean())
print("count:", sup["보충"].count())
print("len  :", len(sup))
```
