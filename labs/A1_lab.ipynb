## Part A. 1학기 방식 — 딕셔너리로 만든 학생

지난 학기까지 배운 것으로 학생을 표현하면 이렇습니다. 이 셀은 **그대로 실행만 하세요.**

> 평균은 아직 내지 않습니다 — **Part E에서 손으로 먼저** 구할 것이기 때문입니다.

```python
def make_student(name, dept):
    return {"name": name, "dept": dept, "scores": []}

def add_score(student, point):
    student["scores"].append(point)

def average(student):
    return round(sum(student["scores"]) / len(student["scores"]), 1)

subin_d = make_student("강수빈", "컴퓨터")
minjun_d = make_student("노민준", "경영")
jiwoo_d = make_student("서지우", "컴퓨터")

for p in [92, 87, 88]:
    add_score(subin_d, p)
for p in [78, 82, 80]:
    add_score(minjun_d, p)
for p in [65, 70, 75]:
    add_score(jiwoo_d, p)

for s in [subin_d, minjun_d, jiwoo_d]:
    print(s)
```

## Part I. 클래스 속성이 옳은 자리

**정말로 모두가 같이 쓸 값**이라면 클래스 본문이 옳은 자리입니다.
판단 기준은 하나 — *이 값이 인스턴스마다 달라야 하는가?*

```python
class Student:
    school = "한빛대학교"
    made = 0

    def __init__(self, name, dept):
        self.name = name
        self.dept = dept
        self.scores = []
        Student.made = Student.made + 1

    def average(self):
        return round(sum(self.scores) / len(self.scores), 1)

a = Student("강수빈", "컴퓨터")
b = Student("노민준", "경영")
c = Student("서지우", "컴퓨터")

print(a.school, b.school, c.school)
print("지금까지 만든 학생 수:", Student.made)
print("이름은 각자:", a.name, b.name, c.name)
```

## Part J. 오류 읽기 — 클래스를 처음 만들 때

일부러 네 가지 오류를 내 봅니다. **메시지의 앞부분**을 읽는 연습입니다.
여기서는 `try/except`로 **잡아서 찍기** 때문에 메시지만 나옵니다.
잡지 않아 **빨간 글씨로 나올 때는** `AttributeError` 뒤에 "혹시 이건가요?" 같은 제안이 붙기도 합니다(버전에 따라 다릅니다).

```python
try:
    Student("강수빈")
except TypeError as e:
    print("1)", type(e).__name__, "->", e)

try:
    print(a.score)
except AttributeError as e:
    print("2)", type(e).__name__, "->", e)

class Wrong:
    def __init__(self, name):
        self.name = name

    def hello():
        return "안녕"

try:
    Wrong("가").hello()
except TypeError as e:
    print("3)", type(e).__name__, "->", e)

class Wrong2:
    def __init__(self, name):
        self.name = name

    def hello(self):
        return "안녕 " + name

try:
    Wrong2("나").hello()
except NameError as e:
    print("4)", type(e).__name__, "->", e)
```
