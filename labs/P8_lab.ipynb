## Part B. 오류 도감 — 이름만 보고 알아맞히기

여섯 가지를 일부러 내 보고 **이름과 메시지**를 확인합니다.


````python
bad_codes = [
    "print(scoers)",          # 없는 이름
    "'3' + 1",                # 글자에 숫자를 더하기
    "int('66점')",            # 숫자가 아닌 글자를 숫자로
    "[72, 55, 91][3]",        # 없는 자리
    "{'a': 1}['b']",          # 없는 열쇠
    "10 / 0",                 # 0 으로 나누기
]

names = []
for code in bad_codes:
    try:
        eval(code)
    except Exception as e:
        names.append(type(e).__name__)
        print(f"{code:<22} {type(e).__name__}: {e}")

print(names)
````

그래서 트레이스백의 **첫 칸은 13보다 큰 번호**로 나옵니다 — 나머지 두 칸은 위 번호와 같습니다.

**고치지 말고 그냥 실행해 보세요.**


````python
def average(nums):
    return sum(nums) / len(nums)

def top_scores(nums, cut):
    for n in nums:
        picked = []
        if n > cut:
            picked.append(n)
    return picked

def report(nums, cut):
    best = top_scores(nums, cut)
    return average(best)

scores = [72, 55, 91, 68, 84, 47]

try:
    print(report(scores, 70))
except ZeroDivisionError:
    print(traceback.format_exc())
````
