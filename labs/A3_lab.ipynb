```python
STATS = '''def mean(nums):
    return sum(nums) / len(nums)

def median(nums):
    return nums[len(nums) // 2]
'''

with open("statistics.py", "w", encoding="utf-8") as f:
    f.write(STATS)

import statistics

mean_now = statistics.mean(scores)
wrong_median = statistics.median(scores)

print("평균  :", mean_now)
print("중앙값:", wrong_median)
print("손으로 구한 중앙값:", med_hand)
```

```python
FIXED = '''def mean(nums):
    return sum(nums) / len(nums)

def median(nums):
    return sorted(nums)[len(nums) // 2]
'''

with open("statistics.py", "w", encoding="utf-8") as f:
    f.write(FIXED)

import statistics
after_cache = statistics.median(scores)
print("파일을 고치고 다시 import →", after_cache)

del sys.___["statistics"]
import statistics
after_reload = statistics.median(scores)
print("담아 둔 것을 지우고 다시 import →", after_reload)
```

## Part J. 패키지 — 폴더로 묶기

파일이 여러 개면 **폴더로 묶습니다.** 폴더 안에 `__init__.py`를 두면 그 폴더가 **패키지**가 됩니다.

`gradepack` 폴더에 **두 파일**을 담습니다 — 함수들이 든 파일과, **A1·A2에서 만든 `Student` 클래스**가 든 파일.
클래스도 이렇게 파일로 옮겨 두면 어디서나 불러 쓸 수 있습니다.

> `os.path.join`은 폴더 이름과 파일 이름을 이어 주고, `open(...).close()`는 **빈 파일 하나**를 만듭니다.
>
> **빈칸 7** — `mean`이 들어 있는 쪽은 **두 파일 중 어느 것**일까요? 점(`.`)이 **폴더 구분**입니다.

```python
STUDENT = '''class Student:
    def __init__(self, name, dept):
        self.name = name
        self.dept = dept
        self.scores = []

    def add(self, score):
        self.scores.append(score)

    def average(self):
        return sum(self.scores) / len(self.scores)
'''

os.makedirs("gradepack", exist_ok=True)
open(os.path.join("gradepack", "__init__.py"), "w", encoding="utf-8").close()

for fname, src in (("grades.py", FIXED), ("student.py", STUDENT)):
    with open(os.path.join("gradepack", fname), "w", encoding="utf-8") as f:
        f.write(src)

from gradepack.___ import mean
from gradepack.student import Student

pkg_mean = mean(scores)
subin = Student("강수빈", "컴퓨터")
subin.add(89)

print("패키지에서 가져온 mean:", pkg_mean)
print("클래스도 파일로 옮겨 왔습니다:", subin.name, subin.average())
```
