## Part D. 열쇠의 자격 — P5의 불변 회수

문자열·숫자·튜플은 열쇠가 됩니다. 그런데 **리스트는 안 됩니다.**
왜 안 되는지 메시지를 읽어 보세요.

````python
d = {}
d["python"] = 1            # 문자열 열쇠
d[(2026, 9)] = "튜플"       # 튜플 열쇠
print(d)
try:
    d[["python"]] = 1      # 리스트를 열쇠로
except TypeError as e:
    print("TypeError:", e)
key_kinds = len(d)         # 성공한 열쇠는 몇 개인가
print(key_kinds)
````
