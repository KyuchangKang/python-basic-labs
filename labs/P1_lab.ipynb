## Part C. 변수 — 값에 이름표를 붙인다

`=`는 "같다"가 아니라 **"오른쪽을 계산해서 왼쪽 이름에 붙여라"라는 명령**입니다.
아래에서 `y = x` 다음에 `x`를 바꿔도 `y`가 그대로인지 확인하세요.

````python
x = 100
y = x              # 값 100에 y 라는 이름표를 하나 더 붙인다
x = 200            # x 이름표만 값 200으로 옮겨 간다
print("x =", x)    # 200
print("y =", y)    # 100 — y는 그대로

count = 0
count = count + 1  # 오른쪽을 먼저 계산 → 그 결과에 count 이름을 다시 붙인다
count = count + 1
print("count =", count)
````

## Part D. 세트피스 — 세 자리 돌려 앉기 ⭐

`a=10, b=20, c=30`을 **오른쪽으로 한 칸 밀어** `(30, 10, 20)`으로 만드는 것이 목표입니다.

**⚠️ 코드를 실행하기 전에, 아래 표를 손으로 먼저 채워 보세요.**

| 실행한 줄 | a | b | c |
|---|---|---|---|
| (시작) | 10 | 20 | 30 |
| `a = c` |  |  |  |
| `b = a` |  |  |  |
| `c = b` |  |  |  |

다 채웠으면 실행해서 맞는지 확인합니다.

````python
a, b, c = 10, 20, 30
print("시작      :", (a, b, c))

a = c
print("a = c 후  :", (a, b, c))    # a의 원래 값 10이 사라졌다
b = a
print("b = a 후  :", (a, b, c))    # 이 줄의 a는 이미 30!
c = b
print("c = b 후  :", (a, b, c))

print("순진한 시도 결과:", (a, b, c), "— 원하던 (30, 10, 20)이 아니다")
````

> **왜 실패했나.** 2행 `b = a`를 쓸 때 내 머릿속의 `a`는 "원래의 10"이지만,
> 파이썬이 2행에 도착한 시점의 `a`는 1행에서 **이미 30으로 덮어써진 값**입니다.
> **대입은 동시에 일어나지 않습니다. 위에서 아래로, 한 줄씩입니다.**

### 고치는 방법 ① — 임시 변수

값을 덮어쓰기 **전에** 다른 이름으로 대피시켜 둡니다.

````python
a, b, c = 10, 20, 30
tmp = a           # 10을 tmp에 대피시킨다
a = c             # a는 30
c = b             # c는 20
b = tmp           # b는 대피시켜 둔 10
print("임시 변수 방법:", (a, b, c))

result_tmp = (a, b, c)   # 뒤에서 두 방법을 비교하려고 저장해 둔다
````

### ✍️ 빈칸 3·4·5 — 고치는 방법 ② 파이썬의 동시 대입

파이썬은 **오른쪽을 먼저 전부 계산한 뒤** 왼쪽에 한꺼번에 붙입니다.
`a`에는 `c`의 값, `b`에는 `a`의 값, `c`에는 `b`의 값이 가도록 오른쪽 세 칸을 채우세요.

````python
a, b, c = 10, 20, 30
a, b, c = ___, ___, ___    # 오른쪽을 먼저 전부 계산 → 그다음 한꺼번에 대입
print("동시 대입 방법:", (a, b, c))

result_swap = (a, b, c)
````

````python
print("임시 변수 :", result_tmp)
print("동시 대입 :", result_swap)
print("두 방법이 같은가?", result_tmp == result_swap)

assert result_swap == (30, 10, 20), "목표는 (30, 10, 20) 입니다"
print("세트피스 검산 통과")
````

## Part F. 출력 다듬기 — f-string

### ✍️ 빈칸 7 — 따옴표 바로 앞에 붙일 **한 글자**를 채우세요.

````python
name = "민수"
age = 20

print(name + "님은 " + str(age) + "살입니다.")           # str()을 매번 붙여야 하는 방법
msg = ___"{name}님은 {age}살, 10년 뒤엔 {age + 10}살입니다."
print(msg)
print(f"10 나누기 3 = {10 / 3:.2f}")                     # 소수점 둘째 자리까지
````

아래 셀은 오류 3종을 **일부러** 내고 메시지만 받아 출력합니다.
(Part E에서 본 `try` / `except` — 오류를 붙잡아 메시지만 꺼내는 장치입니다. 문법은 P8에서 배웁니다.)

**메시지를 소리 내어 읽어 보세요.** 영어라서 그냥 넘기지 말고, 다음 세 조각만 짚으면 됩니다:
`is not defined`(정의된 적 없다) · `can only concatenate str`(문자열끼리만 이어 붙일 수 있다) · `invalid literal`(유효하지 않은 표기)

````python
try:
    print(nickname)              # 만든 적 없는 이름을 부르면?
except NameError as e:
    print("NameError  :", e)

try:
    "나이: " + 20                # 문자열에 숫자를 이어 붙이면?
except TypeError as e:
    print("TypeError  :", e)

try:
    int("스물")                  # 숫자로 바꿀 수 없는 글자를 int로 바꾸면?
except ValueError as e:
    print("ValueError :", e)
````

## Part H. `assert`로 스스로 채점하기

`assert 조건`은 조건이 참이면 조용히 지나가고, 거짓이면 오류를 냅니다.
아래 셀이 **아무 오류 없이 마지막 줄을 출력하면** 오늘 실습을 제대로 채운 것입니다.

````python
assert 3 + 4 == 7                                    # 숫자 덧셈
assert joined == "34"                                # 빈칸 1·2 — 문자열은 이어 붙는다
assert type(3.0) == float                            # 소수점이 붙으면 float
assert result_swap == (30, 10, 20)                   # 빈칸 3·4·5 — 세트피스 성공
assert result_tmp == result_swap                     # 두 방법의 결과가 같다
assert age_num == 25 and type(age_num) == int        # 빈칸 6 — float 말고 int 로 바꿨나
assert msg == "민수님은 20살, 10년 뒤엔 30살입니다."      # 빈칸 7 — f 를 붙였나
assert int(3.9) == 3                                 # 반올림이 아니라 버림

print("전부 통과 — 수고하셨습니다")
````
