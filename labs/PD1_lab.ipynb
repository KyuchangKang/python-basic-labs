## Part D. 리스트의 `+` 와 Series의 `+`

**같은 기호가 전혀 다른 일을 합니다.** 리스트로는 안 되던 것도 해 봅니다.

## Part F. 세트피스 ⭐ — 두 Series를 더합니다

받은 그대로 Series를 만들어 더합니다. **아주 평범한 코드입니다.**

```python
mid_plain = pd.Series([88, 76, 82, 87, 83])
fin_plain = pd.Series([92, 91, 89, 87, 80])

total_plain = mid_plain + fin_plain
print(total_plain)
print("합계:", total_plain.sum(), "/ 평균:", round(total_plain.mean(), 1))
```

## Part G. 원인 찾기 — 무엇과 무엇을 더했나

`index=` 를 주지 않았으므로 두 Series의 이름표는 **둘 다 `0, 1, 2, 3, 4`** 입니다.
판다스는 **같은 이름표끼리** 더합니다.

| 자리 | 중간의 그 자리 | 기말의 그 자리 | 더한 값 |
|---|---|---|---|
| 0 | 강수빈 88 | **정하람** 92 | 180 |
| 1 | 노민준 76 | **오하린** 91 | 167 |
| 2 | 정하람 82 | **강수빈** 89 | 171 |
| 3 | 오하린 87 | 한도윤 87 | 174 |
| 4 | 한도윤 83 | **노민준** 80 | 163 |

**더한 다섯 숫자는 그대로**이므로 합계와 평균은 **당연히** 맞습니다 — 순서만 섞였으니까요.

> **합계가 맞는다는 것은 "같은 숫자들을 더했다"는 뜻이지 "제대로 짝지었다"는 뜻이 아닙니다.**

```python
print("중간의 이름표:", list(mid_plain.index))
print("기말의 이름표:", list(fin_plain.index))
print("이름표가 같은가?", list(mid_plain.index) == list(fin_plain.index))
print("강수빈의 진짜 총점:", 88 + 89, "/ 나온 값:", total_plain.iloc[0])
```

## Part J. 오류 읽기 — Series를 다룰 때

일부러 세 가지 오류를 내 봅니다. **메시지의 앞부분**을 읽는 연습입니다.

```python
try:
    print(mid2["문서준"])
except KeyError as e:
    print("1)", type(e).__name__, "->", e)

try:
    print([1, 2, 3] + 5)
except TypeError as e:
    print("2)", type(e).__name__, "->", e)

try:
    print(mid2.iloc["강수빈"])
except TypeError as e:
    print("3)", type(e).__name__, "->", e)
```
