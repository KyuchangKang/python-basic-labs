## Part A. 지역 이름 — 함수 안에서 만든 이름

함수 안의 `n`을 함수 밖에서 찍어 보면 어떻게 될까요?

📌 빈칸 1 — 함수 밖에서 `n`이 **보였나요?** `True` 또는 `False`로 답하세요.


````python
scores = [92, 58, 45, 63, 88, 66, 51, 60]

def count_pass(nums, cut):
    n = 0                          # 함수 안에서 만든 이름
    for s in nums:
        if s >= cut:
            n = n + 1
    return n

print(count_pass(scores, 60))      # 기준 60 이면 몇 명인가

try:
    print(n)                       # 함수 밖에서 n 을 찍어 보면?
    n_visible = True
except NameError as e:
    print("NameError:", e)
    n_visible = ___                # 빈칸 1 — 밖에서 보였나
````

## Part B. 전역 이름 — 밖의 이름은 안에서 읽힌다

함수는 **부를 때의** 밖 값을 봅니다. 밖에서 값을 바꾸고 다시 불러 확인하세요.
`msg`는 함수 안에서만 사는 이름이고, `cut`은 밖에서 가져다 쓰는 이름입니다.


````python
cut = 60                       # 밖에서 만든 이름 (전역)

def show_cut():
    msg = "함수 안에서 본 cut:"   # 안에서만 사는 이름
    print(msg, cut)              # cut 은 안에 없으니 밖을 본다
    return cut

seen_before = show_cut()
cut = 75                       # 밖에서 바꾼다
seen_after = show_cut()        # 다시 부르면?
print(seen_before, seen_after)

cut = 60                       # 세트피스를 위해 되돌려 놓는다
````

## Part C. 세트피스 — 기준을 70으로 올리기 ⭐

아래는 기준을 바꾸는 함수입니다. **고치지 말고 그냥 실행해 보세요.**


````python
def set_cut(new_cut):
    cut = new_cut
    print("함수 안:", cut)

set_cut(70)
print("함수 밖:", cut)

wrong_passed = 0
for s in scores:
    if s >= cut:
        wrong_passed = wrong_passed + 1
print("합격자:", wrong_passed)
````
