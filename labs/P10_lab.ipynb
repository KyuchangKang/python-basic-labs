## Part H. `csv` 모듈로 검산

`csv` 모듈은 줄바꿈을 알아서 떼 줍니다. 세트피스의 버그가 **아예 생기지 않습니다.**

> **빈칸 7** — CSV를 **읽는** 도구의 이름을 쓰세요.


```python
import csv

with open("students.csv", encoding="utf-8") as f:
    rows2 = list(csv.___(f))

print(rows2[0])
print(rows2[1])

picked = [int(r[1]) for r in rows2[1:] if r[2] == "3학년"]
csv_avg = round(sum(picked) / len(picked), 1)
print(len(picked), "명 · 평균", csv_avg)
```

### 그런데 왜 모듈까지 필요한가 — 값 안의 쉼표

주소처럼 **값 안에 쉼표가 든** 경우 `split(",")`은 깨집니다. CSV는 그럴 때 큰따옴표로 묶는 것이 약속인데,
`split`은 그 약속을 모릅니다.


```python
import io

one = '노민준,"부산 해운대구, 좌동",78'    # 작은따옴표로 감싸면 안에 큰따옴표를 쓸 수 있다
print(one)
print("split :", one.split(","), "→", len(one.split(",")), "칸")
print("csv   :", list(csv.reader(io.StringIO(one))))   # StringIO 는 문자열을 파일처럼 넘겨 주는 도구
```

## Part I. 파일 쓰기와 뒷정리

3학년 명단을 새 파일로 저장해 봅니다. 🔴 **`"w"`는 기존 내용을 통째로 지웁니다**(이어 쓰려면 `"a"`).

Part A에서 파일을 만들 때 쓴 것이 이 모드입니다. 그때 `f.write("이름,점수,학년@NL@")`처럼 @NL@을 직접 적었지요 —
**`f.write()`는 `print`와 달리 줄바꿈을 자동으로 넣지 않기 때문**입니다. 세트피스의 그 줄바꿈을 이번엔 **내가** 넣은 것입니다.
아래는 `csv.writer`로 쓰는 방법입니다. `newline=""`은 `csv.writer`를 쓸 때만 붙입니다(윈도우에서 빈 줄이 끼는 것을 막습니다).


```python
with open("result.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["이름", "점수"])
    for r in rows2[1:]:
        if r[2] == "3학년":
            writer.writerow([r[0], r[1]])

with open("result.csv", encoding="utf-8") as f:
    saved = list(csv.reader(f))
print(saved)
print("저장된 학생 수:", len(saved) - 1)
```
