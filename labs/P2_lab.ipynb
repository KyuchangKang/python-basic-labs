## Part A. 비교 연산자 — `bool`을 만드는 공장

````python
print(3 == 3)        # 같다
print(3 != 3)        # 다르다
print(3 < 5)         # 작다
print(89.5 >= 90)    # 이상 — 오늘의 경계값 미리보기

result = 3 < 5       # 비교 결과에 이름을 붙일 수도 있다
print(result, type(result))
````

## Part C. `if` / `else` 와 들여쓰기

**들여쓰기는 장식이 아니라 문법입니다.** 아래 두 `print`의 차이를 보세요.

````python
score = 95

if score >= 90:
    print("합격")        # 들여쓰기됨 → 조건이 참일 때만
else:
    print("불합격")
print("판정 끝")          # 들여쓰기 안 됨 → 항상 실행

score = 40               # 점수를 바꿔서 다시 확인
if score >= 90:
    print("합격")
else:
    print("불합격")
print("판정 끝")          # 이번에도 실행된다
````

## Part D. 세트피스 — 다섯 학생의 등급 판정 ⭐

**기준:** 90 이상 A · 80 이상 B · 70 이상 C · 60 이상 D · 그 미만 F

| 학생 | 가영 | 나은 | 다인 | 라온 | 마루 |
|---|---|---|---|---|---|
| 점수 | 95 | 89.5 | 72 | 60 | 45 |

**⚠️ 실행하기 전에, 가영(95점)을 손으로 추적해 아래 표를 채우세요.**

| 줄 | 조건이 참? | 실행 후 `grade` |
|---|---|---|
| (시작) | | `"F"` |
| `if 95 >= 90` | | |
| `if 95 >= 80` | | |
| `if 95 >= 70` | | |
| `if 95 >= 60` | | |

다 채웠으면 실행해서 맞는지 확인합니다.

````python
score = 95
grade = "F"

if score >= 90: grade = "A"
print("90 이상 검사 후:", grade)
if score >= 80: grade = "B"
print("80 이상 검사 후:", grade)    # 참이라 덮어썼다
if score >= 70: grade = "C"
print("70 이상 검사 후:", grade)
if score >= 60: grade = "D"
print("60 이상 검사 후:", grade)

print("실수 1 최종 결과:", grade, "— 95점인데!")
````

> **실수 ①의 원인:** 네 조건이 **전부 참**이라 네 번 다 실행되고, **마지막에 참인 것이 이겼습니다.**
> P1의 세 자리 돌려 앉기와 같은 이야기 — 나중 줄이 앞 줄을 덮어씁니다.

이제 세 가지 판정 방법을 각각 만들어 다섯 학생 전체에 돌려 봅니다.

> **📌 아직 안 배운 문법 — 지금은 몰라도 됩니다.**
> 아래 셀의 `def` 는 **코드 묶음에 이름을 붙여 두는 장치**이고 `return` 은 **그 결과를 돌려주는 것**입니다.
> 판정 방법 세 개를 이름 붙여 두었다가 다섯 학생에게 돌려 보려고 쓴 것뿐입니다. **9주차(P7a)에 제대로 배웁니다.**
> 지금 볼 것은 `def` 줄 **안쪽의 `if`/`elif` 뿐**입니다.

````python
def wrong_many_if(s):        # 실수 1 — if 를 네 개 나열
    g = "F"
    if s >= 90: g = "A"
    if s >= 80: g = "B"
    if s >= 70: g = "C"
    if s >= 60: g = "D"
    return g

def wrong_order(s):          # 실수 2 — elif 를 오름차순으로
    if s >= 60: return "D"
    elif s >= 70: return "C"
    elif s >= 80: return "B"
    elif s >= 90: return "A"
    else: return "F"

print("실수 1, 95점 :", wrong_many_if(95))
print("실수 2, 95점 :", wrong_order(95))
````

### ✍️ 빈칸 2·3·4 — 올바른 판정을 완성하세요

`elif` 사슬은 **위에 오는 조건이 먼저 잡아갑니다.**
그러니 **좁은(까다로운) 조건부터** 내려오도록 세 기준 점수를 채우세요.
(90 이상 A · 80 이상 B · 70 이상 C · 60 이상 D · 그 미만 F)

> 💡 확인은 **95점 하나로 하지 마세요.** 95점은 잘못 채워도 A가 나올 수 있습니다
> (첫 줄 `s >= 90`은 빈칸이 아니니까요). 아래에서 **72점과 60점까지** 함께 찍어 봅니다.

````python
def correct(s):              # 올바른 판정 — 좁은 조건부터 내려온다
    if s >= 90: return "A"
    elif s >= ___: return "B"
    elif s >= ___: return "C"
    elif s >= ___: return "D"
    else: return "F"

print("95점:", correct(95), "· 72점:", correct(72), "· 60점:", correct(60))
````

````python
SCORES = [("가영", 95), ("나은", 89.5), ("다인", 72), ("라온", 60), ("마루", 45)]

print("이름   점수    실수1  실수2  올바름")
for name, s in SCORES:
    print(f"{name}  {s:>5}     {wrong_many_if(s)}      {wrong_order(s)}      {correct(s)}")

many_if_result = [wrong_many_if(s) for _, s in SCORES]
order_result = [wrong_order(s) for _, s in SCORES]
correct_result = [correct(s) for _, s in SCORES]

print()
print("실수 1 :", many_if_result)
print("실수 2 :", order_result)
print("올바름 :", correct_result)
print("두 실수의 결과가 같은가?", many_if_result == order_result)
````

> **오류가 하나도 나지 않았습니다.** 프로그램은 멀쩡히 돌고 답만 틀렸습니다.
> 지난주 오류들은 파이썬이 빨간 글씨로 잡아 줬지만, **이런 버그는 손추적으로만 잡힙니다.**

> **📌 위 셀에도 아직 안 배운 문법이 있습니다 — 지금은 몰라도 됩니다.**
> `SCORES = [(...), (...)]` 는 **값을 여러 개 담아 둔 것**(리스트·튜플 → P4),
> `for name, s in SCORES:` 는 **하나씩 꺼내 반복**하는 것(→ **P3a, 바로 다음 주**),
> `[correct(s) for _, s in SCORES]` 는 **결과를 한 줄로 모으는 표현**(→ P9)입니다.
> 오늘은 "다섯 명에게 차례로 같은 판정을 적용했다"까지만 보면 됩니다.

## Part F. `and` / `or` / `not` 과 범위 확인

## Part G. ⑤확인 — 조건문 오류 3종

오류는 파이썬이 **"여기서 막혔어요"라고 알려 주는 메시지**입니다. **맨 아랫줄부터** 읽으세요.
아래 셀은 잘못된 코드를 **글자로만 만들어** 검사시킵니다. (실행이 아니라 문법 검사라 노트북은 멈추지 않습니다.)

````python
NL = chr(10)                                  # 줄바꿈 문자
bad_codes = [
    ("콜론 빠뜨림", "if 3 > 2" + NL + "    print(1)"),
    ("들여쓰기 없음", "if 3 > 2:" + NL + "print(1)"),
    ("== 대신 =", "x = 5" + NL + "if x = 5:" + NL + "    print(1)"),
]

for label, code in bad_codes:
    try:
        compile(code, "<test>", "exec")
        print(f"{label}: 오류 없음")
    except SyntaxError as e:
        print(f"{label:<14} {type(e).__name__}: {e.msg}")
````

> 세 번째 메시지를 보세요 — 파이썬이 **정답까지 알려 줍니다.**
> 하지만 오늘의 진짜 교훈은 이겁니다: **가장 위험한 것은 오류가 나는 코드가 아니라, 오류 없이 조용히 틀리는 코드**(95점 → D).

## Part H. `assert`로 스스로 채점하기

아래 셀이 **아무 오류 없이 마지막 줄을 출력하면** 오늘 실습을 제대로 채운 것입니다.

````python
assert (3 < 5) is True                                   # 비교식은 bool 값이다
assert correct(95) == "A"                                # 올바른 판정
assert correct(72) == "C" and correct(60) == "D"         # 빈칸 2·3·4 — 95점만으로는 부족하다
assert correct_result == ["A", "B", "C", "D", "F"]       # 다섯 학생 전체
assert many_if_result == ["D", "D", "D", "D", "F"]       # 실수 1 재현
assert order_result == many_if_result                    # 두 실수의 결과가 같다
assert correct(89.5) == "B"                              # 경계값 — 반올림하지 않는다
assert rounded == 90 and correct(rounded) == "A"         # 빈칸 5 — 버림이 아니라 반올림
assert both is False                                     # 빈칸 6 — and
assert way2 is False and way1 == way2                    # 빈칸 7 — 비교 대상은 score

print("전부 통과 — 수고하셨습니다")
````
