## Part F. 빠진 줄을 되살리기

두 가지 길이 있습니다. 하나는 **있는 그대로 보여 주는 것**, 다른 하나는 **빈 칸에 이름을 주는 것**입니다.

```python
with_nan = df.groupby("동아리", dropna=___)["점수"].agg(["count", "mean"])
print(with_nan)
print("인원 합:", with_nan["count"].sum())
```

```python
df["동아리"] = df["동아리"].fillna("미가입")
fixed = df.groupby("동아리")["점수"].agg(["count", "mean"])
print(fixed)
print("인원 합:", fixed["count"].sum())
```

## Part G. 집계 결과도 하나의 표다

`groupby`가 돌려주는 것은 그림이 아니라 **Series 또는 DataFrame**입니다.
PD3에서 배운 `loc`과 정렬이 그대로 통하고, `sort_index`는 **순위를 본 뒤 이름 순서로 되돌릴 때** 씁니다.

```python
dept = df.groupby("학과")["점수"].mean()
print(dept.loc["컴퓨터"])
print(dept.sort_values(ascending=False))    # 값으로 — 순위
print(dept.sort_index())                    # 이름표로 — 원래 순서

print(fixed.reset_index())
print(df.set_index("이름").loc["정하람"])
```

## Part H. 같은 표를 다른 모양으로

`value_counts`는 **세기만** 할 때, `pivot_table`은 열쇠 둘을 **가로세로로 펼** 때 씁니다.
🔴 `pivot_table` 안의 `NaN`은 빈 칸이 아니라 **그 칸에 해당하는 사람이 없다**는 뜻입니다.

```python
print(df["학과"].value_counts())
print(df.pivot_table(index="학과", columns="동아리", values="점수", aggfunc="mean"))
print(df.pivot_table(index="학과", columns="동아리", values="점수", aggfunc="count"))
```
