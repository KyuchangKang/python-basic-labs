## Part A. `for` — 여러 개를 하나씩 꺼낸다

대괄호 `[ ]`로 묶은 것을 **리스트**라고 부릅니다.

````python
temps = [-3, -7, -1, -5, -2]    # 어느 겨울 닷새의 최저기온

for t in temps:                 # temps 에서 하나씩 꺼내 t 라 부른다
    print(t)

print("값 개수:", len(temps))    # 몇 개인지
````

## Part E. 함정 ① — 최댓값 상자를 `0`으로 시작하면

가장 따뜻했던 날을 찾아봅시다. 합계를 `0`으로 시작했으니 최댓값도 `0`으로 시작하면 될 것 같습니다.

**⚠️ 실행하기 전에 예측하세요.** 답이 무엇일 것 같나요?

````python
biggest_bad = 0
for t in temps:
    if t > biggest_bad:
        biggest_bad = t
    print("t =", t, "→ biggest =", biggest_bad)

print("함정 1 결과:", biggest_bad, "← 이 값이 데이터 안에 있나요?")
print("데이터:", temps)
````

> **`0`은 데이터에 없는 값입니다.** 다섯 날 중 0도인 날은 없었는데,
> 프로그램은 "가장 따뜻했던 날은 0도"라고 대답했습니다. **존재하지 않는 날을 발명한 것**입니다.
> P1은 값이 증발했고(`(30,30,30)`), P2는 95점이 D였죠. **셋 다 오류가 나지 않습니다.**

### ✍️ 빈칸 3 — 상자를 **데이터 안의 값**으로 시작하도록 채우세요.

대괄호 `[0]`은 **첫 번째 값**을 꺼냅니다. 그렇다면 **어떤 목록**의 첫 값이어야 할까요?
(대괄호로 하나 꺼내기는 P4에서 제대로 배웁니다.)

````python
biggest = ___[0]               # 데이터의 첫 값으로 시작
for t in temps:
    if t > biggest:
        biggest = t
    print("t =", t, "→ biggest =", biggest)

print("올바른 최댓값:", biggest)
````

## Part F. 함정 ② — 초기화를 루프 안에 넣으면

이번엔 상자를 **어디서** 준비하느냐의 문제입니다. 들여쓰기 한 칸 차이입니다.

**⚠️ 실행하기 전에 예측하세요.** 합계 `-18`이 나올까요?

````python
for t in temps:
    total_bad = 0              # 루프 안에 들여쓰기됨 — 매 바퀴 리셋
    total_bad = total_bad + t
    print("t =", t, "→ total_bad =", total_bad)

print("함정 2 결과:", total_bad, "← 합계가 아니라 마지막 값")
````

> 매 바퀴 상자를 비웠으니 **쌓일 수가 없었습니다.**
> 초기화가 루프 안에 있으면 그건 준비가 아니라 **매번 지우기**입니다.

## Part H. ⑤확인 — 반복문에서 나는 오류

아래 셀은 잘못된 코드를 **글자로만 만들어** 검사하거나, `try`/`except`로 메시지만 꺼냅니다.
(`compile`·`try`/`except`는 P8에서 배웁니다. 지금은 **출력된 메시지만 읽으세요.**)

````python
NL = chr(10)                # 줄바꿈 문자
bad_codes = [
    ("콜론 빠뜨림", "for t in temps" + NL + "    print(t)"),
    ("들여쓰기 없음", "for t in temps:" + NL + "print(t)"),
]
for label, code in bad_codes:
    try:
        compile(code, "<test>", "exec")
        print(f"{label}: 오류 없음")
    except SyntaxError as e:
        print(f"{label:<14} {type(e).__name__}: {e.msg}")

try:
    max([])                    # 빈 목록의 최댓값은?
except ValueError as e:
    print(f"{'빈 목록의 max':<14} ValueError: {e}")
````

> **`max([])`가 왜 오류일까요?** 최댓값은 **데이터 중 하나**여야 하는데,
> 데이터가 없으면 **시작할 값도 없기** 때문입니다 — Part E의 그 이야기입니다.
> `biggest = 0`은 이 오류를 내는 대신 **조용히 `0`을 답으로 주었습니다.** 어느 쪽이 더 위험할까요?
