## Part G. 실행 순서 — 네 경우 (빈칸 없음)

네 방이 **언제 실행되는지**를 순서대로 모아 보는 함수입니다.

```python
def demo(text):
    order = []
    try:
        order.append("try")
        int(text)
    except ValueError:
        order.append("except")
    else:
        order.append("else")
    finally:
        order.append("finally")
    return order

case_ok = demo("50")
case_err = demo("아흔하나")

print("① 오류 없음  :", case_ok)
print("② 오류 잡힘  :", case_err)
```

🔴 **`except`와 `else`가 함께 나온 적이 없습니다** — 둘 중 하나만 실행됩니다.
그리고 **`finally`는 두 경우 다** 마지막에 있습니다.

셋째 경우(**안 잡힌 오류**)와 넷째 경우(**`return`이 있을 때**)도 확인해 봅시다.
아래에서는 기록을 **함수 바깥의 `log`** 에 남깁니다 — 오류가 바깥으로 나가면 함수 안의 값은 **사라지기** 때문입니다.

```python
log = []


def demo_uncaught():
    try:
        log.append("try")
        int("아흔하나")
    except KeyError:
        log.append("except")
    finally:
        log.append("finally")


try:
    demo_uncaught()
except ValueError:
    log.append("바깥에서 잡음")


def demo_return():
    try:
        return "try 의 return"
    finally:
        log.append("finally 가 먼저")


value = demo_return()

print("③ 안 잡힌 오류:", log[:3])
print("④ return 이 있을 때:", log[3], "->", value)
```

```python
class ScoreError(Exception):
    pass


def check(score):
    if score < 0 or score > 100:
        ___ ScoreError("점수 범위를 벗어났습니다: " + str(score))
    return score


caught = ""
print("정상:", check(89))

try:
    check(150)
except ScoreError as e:
    caught = str(e)

print("잡은 메시지:", caught)
```

## Part J. 오류 읽기 — 예외를 다룰 때

일부러 세 가지를 내 봅니다. **메시지의 앞부분**을 읽는 연습입니다.
`type(e).__name__` 은 **오류의 이름만 꺼내 보는 장치**입니다(P8 노트북에서 쓰던 그것).

```python
try:
    value = int("아흔하나")
except ValueError as e:
    print("1)", type(e).__name__, "->", e)

try:
    print(100 / 0)
except ZeroDivisionError as e:
    print("2)", type(e).__name__, "->", e)

try:
    scores = {"강수빈": 89}
    print(scores["오하린"])
except KeyError as e:
    print("3)", type(e).__name__, "->", e)
```

🔴 세 번째 `KeyError`는 **A5에서 다시 만납니다** — 겹겹이 쌓인 데이터에서 값을 꺼낼 때입니다.
