## Part A. 오늘의 표 셋을 만든다

아래 셀이 파일 셋을 만듭니다. 6월 1~15일 대여건수, 6월 16~30일 대여건수, 그리고 6월 기상 관측입니다.
설명서 §15.1에 **같은 코드**가 실려 있습니다.

🔴 이 셀에는 **실제 파일에서 흔히 겪는 잡티를 심는 줄**이 하나 들어 있습니다.
**코드를 읽지 말고 그냥 실행한 뒤 Part C에서 직접 찾아보세요.** 미리 읽으면 오늘의 재미가 반으로 줄어듭니다.

🔴 앞 파일의 대여건수는 **쉼표가 든 글자**로, 뒤 파일은 **쉼표 없는 숫자**로 적혀 있습니다.
같은 한 달치인데 **내려받은 시점 사이에 형식이 바뀐** 것입니다.

```python
날짜 = ["2024-06-01", "2024-06-02", "2024-06-03", "2024-06-04", "2024-06-05",
       "2024-06-06", "2024-06-07", "2024-06-08", "2024-06-09", "2024-06-10",
       "2024-06-11", "2024-06-12", "2024-06-13", "2024-06-14", "2024-06-15",
       "2024-06-16", "2024-06-17", "2024-06-18", "2024-06-19", "2024-06-20",
       "2024-06-21", "2024-06-22", "2024-06-23", "2024-06-24", "2024-06-25",
       "2024-06-26", "2024-06-27", "2024-06-28", "2024-06-29", "2024-06-30"]
요일들 = ["토", "일", "월", "화", "수", "목", "금"]
요일 = [요일들[i % 7] for i in range(30)]        # 2024년 6월 1일은 토요일
건수앞 = ["14,230", "15,010", "13,120", "13,540", "6,880", "12,760", "13,980", "16,420",
        "15,870", "13,310", "13,650", "9,240", "14,020", "14,890", "16,060"]
건수뒤 = ["5120", "7340", "10210", "4050", "8760", "12480", "5430", "5970",
        "11390", "9880", "11390", "4610", "7890", "13240", "13880"]
기온 = [24.1, 23.5, 22.8, 24.6, 21.2, 23.9, 25.3, 26.0, 26.4, 25.1,
       24.7, 22.9, 25.8, 26.6, 27.2, 23.1, 22.4, 24.0, 21.6, 23.3,
       26.1, 22.0, 21.8, 25.4, 24.2, 26.8, 22.6, 23.0, 27.5, 27.9]
비온날 = {5: 8.5, 12: 3.2, 16: 12.4, 17: 5.6, 19: 22.8, 20: 3.1,
        22: 15.2, 23: 9.8, 25: 1.2, 27: 18.5, 28: 6.4}
강수 = [비온날.get(d) for d in range(1, 31)]     # 비가 안 온 날은 빈 칸

앞 = pd.DataFrame({"날짜": 날짜[:15], "요일": 요일[:15], "대여건수": 건수앞})
뒤 = pd.DataFrame({"날짜": 날짜[15:], "요일": 요일[15:], "대여건수": 건수뒤})
기상 = pd.DataFrame({"날짜": 날짜, "평균기온": 기온, "강수량": 강수})
기상.loc[18, "날짜"] = " " + 날짜[18]

앞.to_csv("bike_a.csv", index=False, encoding="utf-8")
뒤.to_csv("bike_b.csv", index=False, encoding="utf-8")
기상.to_csv("weather.csv", index=False, encoding="utf-8")

bike_a = pd.read_csv("bike_a.csv")
bike_b = pd.read_csv("bike_b.csv")
weather = pd.read_csv("weather.csv")
print(bike_a.head(2))
print(bike_b.head(2))
print("줄 수:", len(bike_a), len(bike_b), len(weather))
```

## Part B. 두 반달을 한 달로

🔴 표는 **대괄호에 담아 하나로** 넘깁니다. 그리고 `ignore_index=True`로 **번호를 새로 매깁니다**(PD7).

```python
bike = pd.___([bike_a, bike_b], ignore_index=True)

print("붙인 줄 수:", len(bike_a), "+", len(bike_b), "=", len(bike))
print(list(bike.index)[:3], "...", list(bike.index)[-3:])
```

## Part C. 기상 표와 짝짓기

날짜를 열쇠로 두 표를 짝지읍니다. **붙이기 전후로 줄 수를 세는 것**이 PD7에서 배운 첫 검산입니다.

```python
day = pd.merge(bike, weather, on="날짜")

print("자전거", len(bike), "· 기상", len(weather), "· 붙인 결과", len(day))
```

줄 수가 어긋났습니다. **누가 빠졌는지** 찾아봅시다.

- 첫 빈칸: PD3에서 배운 **목록에 있는지 묻는 메서드**입니다. 앞의 `~`와 함께 읽으면 **"기상 표에 날짜가 없는 날만"** 입니다.
- 둘째 줄은 **P10에서 쓴 그 방법**입니다 — 눈에 안 보이는 글자를 **글자 수로** 잡아냅니다(날짜는 열 글자여야 하지요).
- 셋째 빈칸: 문자열 열을 받으면 가장 먼저 하는 일 — **양끝의 공백을 떼는** 메서드입니다(PD5).

```python
missing = bike[~bike["날짜"].___(weather["날짜"])]
print(missing)

print([d for d in weather["날짜"] if len(d) != 10])    # 열 글자가 아닌 날짜 찾기

weather["날짜"] = weather["날짜"].str.___()
day = pd.merge(bike, weather, on="날짜")
print("고친 뒤 붙인 결과:", len(day))
```

## Part D. 세트피스 ⭐ — 6월 하루 평균 대여건수

셋 다 같은 수여야 합니다. 아래 셀의 `hand_days`가 여러분이 손으로 정한 **분모**입니다.

오류가 난 까닭은 `대여건수`가 **글자**이기 때문입니다(PD2).

숫자로 바꿉시다. **바꿀 수 없는 칸을 어떻게 할지**를 `errors=`에 말해 줍니다.
PD2에서 쓴 그 값이고, 뜻은 **"못 바꾸겠으면 빈 칸으로 두라"** 입니다. 따옴표까지 적으세요.

```python
숫자 = pd.to_numeric(day["대여건수"], errors=___)

print("평균:", 숫자.mean())
```

답이 나왔습니다. 손으로 적어 둔 답과 같은가요?

## Part E. 고쳤다고 생각한 순간

쉼표가 문제였으니 쉼표를 떼면 되겠지요. 아래 셀을 실행하고 **값이 있는 칸이 몇 개인지 다시 보세요.**

```python
숫자2 = pd.to_numeric(day["대여건수"].str.replace(",", ""))

print("평균:", 숫자2.mean())
print("줄 수:", len(day), "/ 값이 있는 칸:", len(day) - 숫자2.isna().sum())
print(숫자2.head(3).tolist(), 숫자2.tail(3).tolist())
```

살아남은 쪽이 **아까와 반대**입니다. 붙인 열에 **글자와 숫자가 섞여** 있었기 때문입니다.

`.str`은 글자에만 붙는 도구입니다. 그러니 `.str`을 쓰기 전에 **열 전체를 한 가지로 통일**해야 합니다.
빈칸에는 PD4에서 자료형을 바꿀 때 쓴 그 메서드를 적으세요.

```python
day["대여건수"] = pd.to_numeric(day["대여건수"].___(str).str.replace(",", ""))

print("평균:", day["대여건수"].mean())
print("줄 수:", len(day), "/ 값의 개수:", day["대여건수"].count())
print("합계:", day["대여건수"].sum())
print("최다:", day.loc[day["대여건수"].idxmax(), "날짜"])
```

## Part F. 빈 칸을 어떻게 할 것인가

강수량 열에는 아직 빈 칸이 있습니다. **비가 오지 않은 날**입니다.

- 빈 칸을 그냥 두면 평균은 **비가 온 날의 평균**이 됩니다.
- **6월 하루 평균**을 묻는다면 비가 안 온 날도 분모에 들어가야 합니다.

두 번째 질문에 답하려면 빈 칸을 **0으로 채워야** 합니다. 빈칸에는 그 일을 하는 메서드를 적으세요(PD4).
🔴 PD4에서는 이 도구를 쓰면 **틀렸습니다.** 여기서는 맞습니다 — **판단은 데이터의 뜻이 정합니다.**

```python
print("줄 수:", len(day), "/ 값의 개수:", day["강수량"].count())
print("빈 칸을 뺀 평균:", round(day["강수량"].mean(), 2))

day["강수량"] = day["강수량"].___(0)
print("채운 뒤 평균:", round(day["강수량"].mean(), 2))
```

## Part G. 날씨별로 묶고 그림으로 보이기

강수량으로 **날씨 열**을 만들고 묶습니다. 빈칸에는 PD6에서 배운 **묶는 메서드**를 적으세요.
🔴 묶은 뒤에는 **날 수의 합과 표의 줄 수**를 맞춰 봅니다(PD6의 검산).

이어서 그림입니다. **`matplotlib`은 이 과목에서 처음 나오는 라이브러리**입니다.
파이썬에서 그림을 그리는 표준 도구이고 Colab에 이미 깔려 있습니다. 오늘은 **두 가지만** 씁니다.
🔴 **제목·축 이름은 영어로 적습니다.** Colab 기본 글꼴에 한글이 없어 네모로 깨집니다.

```python
day["날씨"] = day["강수량"].apply(lambda mm: "비" if mm > 0 else "맑음")
표 = day.___("날씨")["대여건수"].agg(["count", "mean"]).round(1)

print(표)
print("날 수 합:", 표["count"].sum(), "/ 줄 수:", len(day))

import matplotlib.pyplot as plt

plt.hist(day["대여건수"], bins=6)
plt.title("Daily Bike Rentals (June)")
plt.xlabel("Rentals")
plt.ylabel("Days")
plt.show()

plt.bar(표.index, 표["mean"])
plt.title("Mean Rentals by Weather")
plt.xlabel("Weather")
plt.ylabel("Mean rentals")
plt.show()
```

## Part H. 기계학습이 받는 모양으로

**보는 것**을 모은 표가 `X`(특징), **맞힐 것** 한 열이 `y`(정답)입니다.

- `X`는 **대괄호 두 겹** — 표로 꺼냅니다.
- `y`는 **대괄호 한 겹** — 한 줄로 꺼냅니다.
- 🔴 맞힐 열을 `X`에 넣으면 안 됩니다. 글자 열도 그대로는 못 넣어 `요일`을 0과 1로 바꿉니다.

```python
day["주말"] = day["요일"].apply(lambda d: 1 if d in ["토", "일"] else 0)

X = day[["평균기온", "강수량", "주말"]]
y = day["대여건수"]
print("X.shape:", X.shape, "/ y.shape:", y.shape)
print(X.head(3))
```
