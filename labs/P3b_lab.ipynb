## Part A. `while` — 조건이 참인 동안

`for`는 꺼낼 값이 정해져 있어야 씁니다. `while`은 **몇 번 돌지 몰라도** 됩니다.

### ✍️ 빈칸 1 — 반복이 끝나도록, 조건에 쓰인 값을 바꾸는 줄을 완성하세요.

````python
n = 5
while n > 0:
    print(n)
    n = n ___ 1          # 조건에 쓰인 n 을 줄여 준다

print("발사!")
print("끝난 뒤 n =", n)
````

## Part B. 무한 루프를 **안전하게** 재현

위 코드에서 `n`을 줄이는 줄을 빼면 어떻게 될까요? **영원히 멈추지 않습니다.**

그대로 실행하면 노트북이 멈추지 않으므로, **10바퀴가 넘으면 `break`로 빠져나오게** 안전장치를 걸었습니다.
(`break`는 바로 다음 Part에서 배웁니다. 여기서는 "강제로 끊는 장치"로만 보세요.)

````python
stuck = 5                # Part A 의 n 과 헷갈리지 않게 다른 이름을 쓴다
guard = 0                # 안전장치 — 몇 바퀴 돌았는지 센다

while stuck > 0:
    guard = guard + 1    # stuck 을 줄이는 줄이 없다!
    if guard >= 10:
        break            # 10바퀴에서 강제로 끊는다

print("10바퀴를 돌았는데도 stuck =", stuck, "— 조건이 계속 참이다")
print("안전장치가 없었다면 영원히 돌았을 것")
````

> **`while`을 쓸 때 반드시 자문할 것: "조건에 쓰인 값을 누가 바꾸나?"**
> `for`는 파이썬이 알아서 다음 값을 꺼내 주지만, **`while`은 내가 직접** 바꿔 줘야 합니다.
>
> 그리고 **무한 루프는 오류가 아닙니다.** 빨간 글씨가 안 나오고 셀이 계속 돌기만 합니다.
> P1·P2·P3a의 함정은 **조용히 틀린 답**을 줬는데, 무한 루프는 **아예 답을 주지 않습니다.**

## Part C. `break` 와 `continue`

### ✍️ 빈칸 2 — 조건을 만족하는 **첫 값을 찾으면 반복을 끝내는** 명령을 채우세요.

````python
temps = [-3, -7, -1, -5, -2]

found = None                 # None = 아직 아무것도 못 찾았다는 표시
checked = 0                  # 몇 개를 확인했는지 — break 면 도중에 멈춘다
for t in temps:
    checked = checked + 1
    print("확인 중:", t)
    if t > -2:
        found = t
        ___                  # 찾았으니 더 볼 필요 없다

print("-2보다 큰 첫 값:", found)
print("확인한 개수:", checked, "/", len(temps), "← 끝까지 안 갔다면 도중에 멈춘 것")
````

````python
kept = []
for t in temps:
    if t < -4:
        continue         # 너무 추운 날은 이번 바퀴만 건너뛴다
    kept.append(t)       # append 는 P4 에서 제대로 — 지금은 모아 둔다는 뜻

print("-4 이상만:", kept)
````

## Part D. 중첩 반복 — 24칸의 실행 순서 ⭐

바깥 3번 × 안쪽 8번 = **24번** 실행됩니다.

**⚠️ 실행하기 전에 예측하세요.** `dan`이 3이 되는 것은 **몇 번째** 실행부터일까요?

> `f"{order:>2}"` 의 `>2`는 **두 칸 폭으로 오른쪽 맞춤**입니다(P1의 `:.2f`와 같은 자리).

````python
order = 0                        # 몇 번째 실행인지 세는 누적기
for dan in [2, 3, 4]:
    for i in range(1, 9):
        order = order + 1
        if i == 1 or i == 8:     # 각 단의 처음과 끝만 찍어 본다
            print(f"{order:>2}번째 실행: dan={dan}, i={i}")

print("총 반복 횟수:", order, "= 3 x 8")
````

> **안쪽 루프가 8번을 다 돌아야 바깥 루프가 한 칸** 갑니다.
> 자동차 주행거리계처럼 — 작은 자리가 한 바퀴 다 돌아야 큰 자리가 1 올라갑니다.

## Part E. `print` 위치가 바꾸는 것

들여쓰기 **한 단계**가 출력 줄 수를 바꿉니다. `dan` 2개 × `i` 3개로 확인합니다.

````python
print("[1] 안쪽 루프 안 — 매 조합마다")
lines_inner = 0
for dan in [2, 3]:
    for i in range(1, 4):
        lines_inner = lines_inner + 1
        print("   ", dan, "x", i)        # 실제로 찍어 본다
print("   줄 수:", lines_inner)

print("[2] 안쪽 밖, 바깥 안 — 단마다 한 번")
lines_outer = 0
for dan in [2, 3]:
    for i in range(1, 4):
        pass                             # pass = 아무것도 하지 않음 (자리만 채운다)
    lines_outer = lines_outer + 1
    print("   ", dan, "단 끝")
print("   줄 수:", lines_outer)

print("[3] 둘 다 밖 — 전부 끝난 뒤 한 번")
print("    전부 끝")
print("   줄 수: 1")
````

## Part F. 세트피스 — 구구단 단별 합계 ⭐

**2단·3단·4단의 단별 합계**를 구합니다. 각 단은 1부터 9까지 곱한 값을 다 더합니다.

**손으로 먼저 구해 보세요.** 1부터 9까지 더하면 45입니다.

| 단 | 계산 | 손으로 구한 합계 |
|---|---|---|
| 2단 | 2 × 45 | |
| 3단 | 3 × 45 | |
| 4단 | 4 × 45 | |

다 채웠으면, **지난주에 배운 대로** 코드를 써 봅니다 —
*"누적기는 루프 밖에서 한 번 준비하고, 루프 안에서는 고쳐 쓰기만."*

````python
total = 0                          # 루프 밖에서 한 번 — 지난주 규칙대로
wrong = []

for dan in [2, 3, 4]:
    for i in range(1, 10):
        total = total + dan * i
    print(dan, "단 합계:", total)
    wrong.append(total)

print("결과:", wrong, "← 손으로 구한 답과 같나요?")
````

> **2단은 90으로 맞았는데 3단이 225입니다.** 정답은 135인데요.
> `total`을 **한 번도 비우지 않아서** 3단을 돌 때 이미 90에서 시작했기 때문입니다.
>
> 사실 `90 → 225 → 405`는 **누적 합계**입니다. 틀린 게 아니라 **다른 것을 구한 것**이죠.

## Part G. 고치기 — 상자를 단마다 새로

### ✍️ 빈칸 3·4 — 누적기를 **단이 바뀔 때마다 새로** 준비하도록 채우세요.

````python
right = []

for dan in [2, 3, 4]:
    total = ___                    # 단이 바뀔 때마다 새 상자
    for i in range(1, 10):
        total = total + dan * ___  # 그 단에서 더할 값
    print(dan, "단 합계:", total)
    right.append(total)

print("결과:", right)
````

### ✍️ 빈칸 5 — 전체 합계(2·3·4단을 다 더한 값)를 구하려면 상자를 어디에?

> `right[0]`은 **첫 번째 값**을 꺼내는 표기입니다 — P4에서 제대로 배웁니다.

````python
grand = 0                          # 전체 합계니까 바깥 루프 밖이 맞다
for dan in [2, 3, 4]:
    for i in range(1, 10):
        grand = grand + dan * i

parts_sum = right[0] + right[1] + ___[2]    # 단별 합계를 더한 값
print("전체 합계:", grand)
print("단별 합계를 더하면:", parts_sum, "← 같은 값이어야 한다")
````

## Part H. ⑤확인 — 반복 심화에서 나는 오류

아래 셀은 잘못된 코드를 **글자로만 만들어** 문법 검사합니다.
(`compile`·`try`/`except`는 P8에서 배웁니다. 지금은 **출력된 메시지만 읽으세요.**)

````python
NL = chr(10)                # 줄바꿈 문자
bad_codes = [
    ("while 콜론 빠뜨림", "while n > 0" + NL + "    n = n - 1"),
    ("중첩 들여쓰기 없음", "for dan in [2, 3]:" + NL + "for i in [1, 2]:" + NL + "    pass"),
]
for label, code in bad_codes:
    try:
        compile(code, "<test>", "exec")
        print(f"{label}: 오류 없음")
    except SyntaxError as e:
        print(f"{label:<18} {type(e).__name__}: {e.msg}")

print("무한 루프          : (오류 없음) — 셀이 계속 돌기만 한다")
````
